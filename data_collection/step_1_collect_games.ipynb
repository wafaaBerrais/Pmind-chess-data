```python
#current player selected
USERNAME="EricRosen"#"AnishGiri" #"Alireza2003"#"Naroditsky"
#user list for scrapping
USERNAME_LIST=["EricRosen","AnishGiri","Alireza2003","Naroditsky"]
#max games number
MAX_GAMES=100000
#allowed speeds
ALLOWED_SPEEDS={"blitz","rapid"}
#get API on local environment variable
API_TOKEN=os.getenv("LICHESS_API_TOKEN")
#min plies (give up less interessing game)
MIN_PLIES=10

ORIGINAL_DATA_PARENT="../data/original"

```
